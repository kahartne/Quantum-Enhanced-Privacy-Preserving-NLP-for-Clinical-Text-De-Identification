```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import numpy as np

#sample input features interchange with data later
x = [0.5, 0.8, 0.3, 0.6]

qc = QuantumCircuit(4)

#Layer 1: put every qubit into superposition
for i in range(4):
    qc.h(i)

#Layer 2: encoding each feature individually
for i in range(4):
    qc.rz(2 * x[i], i)

#Layer 3: encode how each pair of features interact
for i in range(3):
    qc.cx(i, i + 1)
    qc.rz(2 * (np.pi - x[i]) * (np.pi - x[i + 1]), i + 1)
    qc.cx(i, i + 1)

qc.measure_all()

sim = AerSimulator()
result = sim.run(qc, shots=1000).result()
print(result.get_counts())
```

```python
from qiskit.quantum_info import Statevector

#same circuit no measurement to see how x actually shapes the state
qc_zz = QuantumCircuit(2)
qc_zz.h(0)
qc_zz.h(1)
qc_zz.rz(2 * x[0], 0)
qc_zz.rz(2 * x[1], 1)
qc_zz.cx(0, 1)
qc_zz.rz(2 * (np.pi - x[0]) * (np.pi - x[1]), 1)
qc_zz.cx(0, 1)
print("ZZFeatureMap state (2 qubit):", Statevector(qc_zz))
```
