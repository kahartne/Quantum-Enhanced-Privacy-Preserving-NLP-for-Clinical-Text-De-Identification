```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

#2-qubit circuit
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()

sim = AerSimulator()
result = sim.run(qc, shots=1000).result()
print(result.get_counts())
```

```python
from qiskit.quantum_info import Statevector

#quick look at single gates on their own before combining them
qc_x = QuantumCircuit(1)
qc_x.x(0)   #flips 0 to 1
print("X:", Statevector(qc_x))

qc_h = QuantumCircuit(1)
qc_h.h(0)   #creates superposition
print("H:", Statevector(qc_h))

qc_z = QuantumCircuit(1)
qc_z.h(0)
qc_z.z(0)   #flips the phase
print("Z:", Statevector(qc_z))

#Bell state without measuring so I can see the actual state
qc_bell = QuantumCircuit(2)
qc_bell.h(0)
qc_bell.cx(0, 1)
print("\nBell state:", Statevector(qc_bell))
```
